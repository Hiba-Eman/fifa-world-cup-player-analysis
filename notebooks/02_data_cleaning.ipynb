### 1. Imports

```python
import pandas as pd
import numpy as np
```

### 2. Reading data

```python
df = pd.read_csv("../data/players_raw.csv")
```

### 4. Strip extra spaces

```python
df["Player"] = df["Player"].str.strip()
df["Category"] = df["Category"].str.strip()
```

### 5. Clean xG Efficiency

```python
df["xG Efficiency"] = (
    df["xG Efficiency"]
    .str.replace("x", "", regex=False)
)
```

### 6. Fill missing values

```python
numeric_cols = df.select_dtypes(include=["number"]).columns

df[numeric_cols] = df[numeric_cols].fillna(0)
df["xG Efficiency"] = df["xG Efficiency"].fillna("0")
```

### 7. Convert xG Efficiency to numeric

```python
df["xG Efficiency"] = pd.to_numeric(df["xG Efficiency"], errors="coerce").fillna(0)
```

```python
df.isnull().sum()
```

### 8. Creating Goal Contribution

```python
df["Goal Contribution"] = df["Assists"] + df["Attempts At Goal"]
```

### 9. Creating Goal Efficiency

```python
df["Goal Efficiency"] = np.where(
    df["Attempts At Goal"] > 0,
    df["Assists"] / df["Attempts At Goal"],
    0
)
```

### 10.Checking the new columns

```python
df[["Player", "Assists", "Attempts At Goal", "Goal Contribution", "Goal Efficiency"]].head()
```

### 11. Save the cleaned dataset

```python
df.to_csv("../data/players_cleaned.csv", index=False)
```
