### 1. Imports

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import os
```

### 2. Reading data

```python
df = pd.read_csv("../data/players_cleaned.csv")
df.head()
```

### 3. Top 10 Players by Goal Contribution

```python
top10 = df.sort_values("Goal Contribution", ascending=False).head(10)

plt.figure(figsize=(10,6))
plt.bar(top10["Player"], top10["Goal Contribution"], color="coral")
plt.xticks(rotation=45, ha="right")
plt.title("Top 10 Players by Goal Contribution")
plt.xlabel("Player")
plt.ylabel("Goal Contribution")
plt.tight_layout()
plt.show()
```

### 4. Top 10 Players by Assists

```python
top10_assists = df.sort_values("Assists", ascending=False).head(10)

plt.figure(figsize=(10,6))
plt.bar(top10_assists["Player"], top10_assists["Assists"], color="coral")
plt.xticks(rotation=45, ha="right")
plt.title("Top 10 Players by Assists")
plt.xlabel("Player")
plt.ylabel("Assists")
plt.tight_layout()
plt.show()
```

### 5. Distribution of Attempts At Goal (Log Scale)

```python
plt.figure(figsize=(8,5))
plt.hist(df["Attempts At Goal"], bins=10, color="coral")
plt.yscale("log")
plt.title("Distribution of Attempts At Goal (Log Scale)")
plt.xlabel("Attempts At Goal")
plt.ylabel("Number of Players")
plt.tight_layout()
plt.show()
```

### 6. Number of Players by Category

```python
plt.figure(figsize=(10,5))
sns.countplot(data=df, x="Category",color="coral")
plt.xticks(rotation=45)
plt.title("Number of Players by Category")
plt.tight_layout()
plt.show()
```

### 7. Goal Contribution Boxplot

```python
plt.figure(figsize=(8,5))
sns.boxplot(y=df["Goal Contribution"])
plt.title("Goal Contribution Boxplot")
plt.show()
```

### 8. Top Speed vs Goal Contribution

```python
plt.figure(figsize=(8,5))
plt.scatter(df["Top Speed (km/h)"], df["Goal Contribution"], color="coral")
plt.title("Top Speed vs Goal Contribution")
plt.xlabel("Top Speed (km/h)")
plt.ylabel("Goal Contribution")
plt.tight_layout()
plt.show()
```

### 9. Passes vs Passing Accuracy

```python
plt.figure(figsize=(8,5))
plt.scatter(df["Passes"], df["Passing Accuracy (%)"], color="coral")
plt.title("Passes vs Passing Accuracy")
plt.xlabel("Passes")
plt.ylabel("Passing Accuracy (%)")
plt.tight_layout()
plt.show()
```

### 10. Correlation Heatmap

```python
key_stats = [
    "Assists",
    "Attempts At Goal",
    "Passes",
    "Passing Accuracy (%)",
    "Top Speed (km/h)",
    "Total Distance (m)",
    "Goal Contribution",
    "Goal Efficiency"
]

corr = df[key_stats].corr()

plt.figure(figsize=(9,7))
sns.heatmap(corr, annot=True, cmap="coolwarm", fmt=".2f")
plt.title("Correlation Between Key Player Stats")
plt.tight_layout()
plt.show()
```
